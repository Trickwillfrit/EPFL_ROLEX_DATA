```python
import pandas as pd

ITEMS_URL = "https://raw.githubusercontent.com/Trickwillfrit/EPFL_ROLEX_DATA/main/items.csv"
items = pd.read_csv(ITEMS_URL)

# Keep ISBN as text
items["isbn_raw"] = items["ISBN Valid"].astype(str)

# Keep only the first ISBN if several are present
items["isbn_first"] = items["isbn_raw"].str.split(";|,").str[0].str.strip()
items.loc[items["ISBN Valid"].isna(), "isbn_first"] = pd.NA

# Clean ISBN: keep only digits and X/x
items["isbn_clean"] = items["isbn_first"].str.replace(r"[^0-9Xx]", "", regex=True)

# Build Open Library cover URL
def build_cover_url(isbn):
    if pd.isna(isbn) or isbn == "":
        return pd.NA
    return f"https://covers.openlibrary.org/b/isbn/{isbn}-M.jpg?default=false"

items["cover_url"] = items["isbn_clean"].apply(build_cover_url)

items_app = items[[
    "i", "Title", "Author", "Publisher", "Subjects", "ISBN Valid", "isbn_clean", "cover_url"
]].copy()

items_app.to_csv("items_app.csv", index=False)
print("items_app.csv saved successfully.")
display(pd.read_csv("items_app.csv").head())
```
